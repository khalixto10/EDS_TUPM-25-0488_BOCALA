```python
import pandas as pd
import numpy as np

print("[INFO] Generating mock 'turbine_5yr_complex_data.csv' for pipeline testing...")

# Set random seed for reproducible results close to your paper's statistics
np.random.seed(42)
num_samples = 2000

# Generate a continuous time series for March 2026
timestamps = pd.date_range(start="2026-03-01 00:00:00", periods=num_samples, freq="15T")

# Create mock data mimicking a right-skewed log-normal distribution for vibration
# Target metrics from paper: Mean ~2.41, Median ~2.38
vibration_data = np.random.lognormal(mean=0.85, sigma=0.25, size=num_samples) + 0.1

# Generate related parameters matching correlation principles (r = 0.87)
generator_rpm = (vibration_data * 500) + np.random.normal(0, 100, num_samples)
generator_rpm = np.clip(generator_rpm, 800, 1800) # Keep within typical RPM range

# Generate other required columns
breath_id = np.random.randint(1000, 5000, size=num_samples)
oil_pressure = (generator_rpm * 0.03) + np.random.normal(0, 5, num_samples)

# Assemble into a DataFrame matching your ingestion requirements
mock_df = pd.DataFrame({
    'timestamp': timestamps,
    'asset_id': ['WTG-08'] * num_samples, # Set to target turbine
    'breath_id': breath_id,
    'generator_rpm': generator_rpm,
    'vibration_u_in': vibration_data,
    'oil_pressure': oil_pressure
})

# Save to the exact file path your pipeline expects
mock_df.to_csv('turbine_5yr_complex_data.csv', index=False)
print("[SUCCESS] Mock file 'turbine_5yr_complex_data.csv' created successfully! You can now run main.py.")
```
